```python
import pandas as pd

df = pd.read_csv('../csv_files/placement.csv')

## How big is dataset?
print(df.shape)

## How does data look like?
print(df.sample(5))

## What is datatype of each column?
print(df.dtypes)

## Are there any missing values?
print(df.isnull().sum())

## How does data look like mathematically?
print(df.describe())

## Are there any duplicates?
print(df.duplicated().sum())

## How is the correlation between different columns?
print(df.corr())
print(df.corr()["iq"])
```
